```python
import numpy as np

np.random.seed(2026)
a = np.random.rand()
b = np.random.randn(25)

np.random.seed(42)
dice = np.random.randint(1, 7, 10000)

print("Random Number:", a)
print("Random Values:", b)
print("Dice Counts:", np.unique(dice, return_counts=True))
print("Six Probability:", np.mean(dice == 6))
print("Double Six:", np.sum((dice[:-1] == 6) & (dice[1:] == 6)))
```

```python
import numpy as np

tmax = np.array([
[18, 20, 26, 32, 38, 39, 35, 34, 33, 30, 25, 19],
[19, 22, 26, 31, 34, 35, 33, 32, 32, 33, 28, 22],
[15, 18, 23, 29, 35, 36, 33, 32, 31, 28, 22, 17],
[18, 22, 29, 35, 39, 39.6, 37, 35, 34, 32, 26, 20],
[10, 14, 20, 26, 32, 36, 35, 33, 30, 26, 19, 12]
])

print("Shape:", tmax.shape)
print("Yearly Average:", tmax.mean(axis=1).round(2))
print("Hottest Month Index:", tmax.argmax(axis=1))
print("Monthly Average:", tmax.mean(axis=0).round(2))
print("Above 35:", tmax[tmax > 35])
```

```python
import numpy as np

months = np.array(['Jul','Aug','Sep','Oct','Nov','Dec','Jan','Feb','Mar','Apr'])

cpi = np.array([
[28.3,27.4,31.4,26.8,29.2,29.7,28.3,23.1,20.7,17.3],
[11.1,9.6,6.9,7.2,4.9,4.1,2.4,1.5,0.7,0.3]
])

print("Average:", cpi.mean(axis=1).round(2))
print("Above 25%:", months[cpi[0] > 25])

drop = cpi[0] - cpi[1]
print("Monthly Drop:", drop.round(1))
print("Biggest Drop:", months[drop.argmax()])

labels = np.where(cpi[1] < 5, "Low",
         np.where(cpi[1] < 15, "Moderate", "High"))
print("FY2025 Labels:", labels)
```

```python
import numpy as np

marks = np.array([35, 45, 28, 50, 32, 60])

print("Original Marks:")
print(marks)

grace_marks = np.where(marks < 40, marks + 5, marks)

print("Marks After Grace:")
print(grace_marks)

print("Passed Students:")
print(np.sum(grace_marks >= 40))
```

```python
import numpy as np

image = np.array([
    [255, 100, 50, 200],
    [150, 255, 0, 100],
    [50, 100, 255, 150],
    [200, 50, 100, 255]
])

print("Original Image:")
print(image)

print("Image Shape:", image.shape)
print("Total Pixels:", image.size)
print("Maximum Pixel Value:", image.max())
print("Minimum Pixel Value:", image.min())
print("Average Pixel Value:", image.mean())

bright_image = image + 10
print("Bright Image:")
print(bright_image)
```
