```python
import pandas as pd
import numpy as np

df = pd.read_csv("pakistan_used_cars.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# A) Missing values
df = df.dropna(subset=["Brand", "Model"])

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna("Unknown")

# B) Formatting
df["Registered City"] = df["Registered City"].str.strip().str.title()

df["Price"] = (df["Price"].astype(str)
               .str.replace("Rs", "", regex=False)
               .str.replace(",", "", regex=False)
               .str.strip())

# C) Data types
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Year"] = pd.to_numeric(df["Year"], errors="coerce").astype("Int64")

# D) Fix spelling
df["Registered City"] = df["Registered City"].replace(
    {"Sukkar": "Sukkur"}
)

df["Model"] = df["Model"].str.replace(
    "Corrolla", "Corolla", regex=False
)

# E) Remove duplicates
df = df.drop_duplicates()

# F) Fix impossible values
df["KMs Driven"] = pd.to_numeric(df["KMs Driven"], errors="coerce")
df.loc[df["KMs Driven"] < 1000, "KMs Driven"] = np.nan

# Remove missing and fake prices
df = df.dropna(subset=["Price"])
df = df[df["Price"] >= 50000]

df["Price"] = df["Price"].astype("int64")

# G) Final check
print("Final Shape:", df.shape)
print(df.isnull().sum())
print(df.dtypes)
```

```python
# a) Price in lakh
df["Price_lakh"] = df["Price"] / 100000
print("4a:")
print(df[["Price", "Price_lakh"]].head())


# b) Age of cars
df["Age"] = 2020 - df["Year"]
print("4b:")
print(df[["Year", "Age"]].head())


# c) Top 5 brands
print("4c:")
print(df["Brand"].value_counts().head(5))


# d) Top 5 cities
print("4d:")
print(df["Registered City"].value_counts().head(5))


# e) Median price by fuel type
print("4e:")
print(df.groupby("Fuel")["Price"].median())


# f) Corolla GLI median price from 2010 to 2018
print("4f:")
corolla = df[df["Model"].str.contains("Corolla GLI", case=False, na=False)]
year_price = corolla[corolla["Year"].between(2010, 2018)].groupby("Year")["Price"].median()
print(year_price)
print("Price change per year:")
print(year_price.diff())


# g) Five most expensive cars
print("4g:")
print(df.nlargest(5, "Price")[["Brand", "Model", "Year", "Price"]])


# h) Cash vs Installment
print("4h:")
print(df.groupby("Transaction Type")["Price"].agg(["count", "median"]))
```
